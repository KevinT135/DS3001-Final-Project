# Virginia Data Centers & Electricity — Loading Raw Data

```python
from pathlib import Path
import urllib.request
import zipfile

import pandas as pd

pd.set_option("display.max_columns", 100)

RAW = Path("data/raw")
RAW.mkdir(parents=True, exist_ok=True)


def download(url, dest):
    if not dest.exists():
        urllib.request.urlretrieve(url, dest)
    return dest
```

## 2. EIA Form 860 — 2025

Source: https://www.eia.gov/electricity/data/eia860/

```python
eia860_zip = download("https://www.eia.gov/electricity/data/eia860/xls/eia8602025.zip", RAW / "eia8602025.zip")
eia860_dir = RAW / "eia860_2025"
if not eia860_dir.exists():
    zipfile.ZipFile(eia860_zip).extractall(eia860_dir)
sorted(p.name for p in eia860_dir.iterdir())
```

```python
gen_sheets = pd.read_excel(eia860_dir / "3_1_Generator_Y2025.xlsx", sheet_name=None, header=1)
gen_raw = pd.concat(
    [df.assign(status_sheet=name) for name, df in gen_sheets.items()],
    ignore_index=True,
)
gen_va_raw = gen_raw[gen_raw["State"] == "VA"]

print(gen_raw.shape, "->", gen_va_raw.shape, "VA generators")
print(gen_va_raw["status_sheet"].value_counts())
gen_va_raw.head()
```

```python
plant_raw = pd.read_excel(eia860_dir / "2___Plant_Y2025.xlsx", header=1)
plant_va_raw = plant_raw[plant_raw["State"] == "VA"]
print(plant_raw.shape, "->", plant_va_raw.shape, "VA plants")
plant_va_raw.head()
```

## 3. Electric Power Monthly — VA sales, revenue, price

Source: https://www.eia.gov/electricity/monthly/

```python
sales_path = download("https://www.eia.gov/electricity/data/eia861m/xls/sales_revenue.xlsx", RAW / "sales_revenue.xlsx")

sales_raw = pd.read_excel(sales_path, sheet_name="Monthly-States", header=[0, 1, 2])

flat_cols = []
sector = None
for top, measure, unit in sales_raw.columns:
    if not str(top).startswith("Unnamed"):
        sector = top
    if str(measure).startswith("Unnamed"):
        flat_cols.append(unit)  # id columns: Year, Month, State, Data Status
    else:
        flat_cols.append(f"{sector}_{measure}_{unit}".replace(" ", "_").replace("/", "_per_"))
sales_raw.columns = flat_cols

sales_va_raw = sales_raw[sales_raw["State"] == "VA"]
print(sales_raw.shape, "->", sales_va_raw.shape, "VA months")
sales_va_raw.head()
```
